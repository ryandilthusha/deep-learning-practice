1. **Memory Referencing vs Deep Copy:**
   - Create `a = np.array([1, 2, 3])`.
   - Create an independent copy `b`.
   - Modify `b[0] = 100` and demonstrate that `a` remains unaffected.



```python
import numpy as np

a = np.array([1, 2, 3])
print('a is: ', a)
```

```python
b = a.copy()

print('b is: ', b)
```

```python
b[0]  = 100

print('b is: ', b)
print('a is: ', a)
```

2. **Vectorized Mathematical Operations:**
   - Take `v = np.array([1, 2, 3, 4])`.
   - Perform element-wise addition (+2), multiplication (*2), squaring, and sine calculation without loops.
   - Perform element-wise addition between `v` and `w = np.array([10, 20, 30, 40])`.

```python
v = np.array([1, 2, 3, 4])

print(v+2)

print(v*2)

print(np.sqrt(v))

print(np.sin(v))
```

```python
w = np.array([10, 20, 30, 40])

v+w
```
