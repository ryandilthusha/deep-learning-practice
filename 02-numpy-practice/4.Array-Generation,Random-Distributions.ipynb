### Exercise 4: Matrix Generators & Sequence Initialization

**Requirements:**
Generate the following arrays without typing the numbers manually:
1. A `(2, 3)` grid populated entirely with zeros cast as 32-bit integers.
2. A `(2, 2)` grid populated entirely with ones.
3. A `(2, 2)` grid filled completely with the number `99`.
4. A `(2, 4)` grid of random continuous floating-point values between `0.0` and `1.0`.
5. A `(2, 4)` grid of random integers bounded between `2` and `10`.
6. An evenly spaced sequential array starting at `2` and ending before `20` with an increment step of `2`.
7. A `5x5` square matrix with ones on the main diagonal and zeros everywhere else.

<details>
<summary>💡 Need a hint?</summary>

- Zeros: `np.zeros(shape, dtype=...)`
- Ones: `np.ones(shape)`
- Fixed constant: `np.full(shape, fill_value)`
- Continuous uniform floats: `np.random.rand(rows, cols)`
- Discrete random integers: `np.random.randint(low, high, size=shape)`
- Stepped sequences: `np.arange(start, stop, step)`
- Identity square: `np.identity(n)`
</details>

Generate `(2, 3)` grid populated entirely with zeros cast as 32-bit integers

```python
import numpy as np

a = np.zeros( (2, 3) , dtype='int32' )
a
```

Generate `(2, 2)` grid populated entirely with ones

```python
b = np.ones( (2, 2) )
b
```

Generate `(2, 2)` grid filled completely with the number `99`

```python
c = np.full( (2, 2) , 99 )
c
```

Generate `(2, 4)` grid of random continuous floating-point values between `0.0` and `1.0`

```python
d = np.random.rand(2,4)
d
```

Generate `(2, 4)` grid of random integers bounded between `2` and `10`

```python
e = np.random.randint( 2, 10, (2,4))
e
```

Generate an evenly spaced sequential array starting at `2` and ending before `20` with an increment step of `2`

```python
f = np.arange(2, 20, 2)
f
```

Generate a `5x5` square matrix with ones on the main diagonal and zeros everywhere else

```python
g = np.identity(5)
g
```
