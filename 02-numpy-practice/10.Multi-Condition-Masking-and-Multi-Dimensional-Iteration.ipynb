1. Generate an array `dataset = np.arange(1, 21).reshape(2, 10)` to simulate imported 2D data.

```python
import numpy as np

dataset = np.arange(1, 21).reshape(2, 10)
dataset
```

2. Extract all elements that are strictly greater than `10`.

```python
a1 = dataset[dataset>10]
a1
```

3. Test column-wise whether **any** value in each column exceeds `15`.

```python
a2 = np.any(dataset>15, axis=0)
a2
```

4. Extract all elements that satisfy two conditions simultaneously: strictly greater than `5` AND strictly less than `10`.

```python
a3 = dataset[((dataset>5) & (dataset>10))]
a3
```
