### Exercise 1: Array Creation & Memory Footprint
1. Create a 1-dimensional array containing `[1, 2, 3]` stored explicitly as 16-bit integers.
2. Create a 2-dimensional array containing `[[1, 2, 3], [4, 5, 6]]` using default integer precision.
3. For both arrays, print:
   - The total number of dimensional axes.
   - The structural dimensions (shape).
   - The internal data type.
   - The memory consumed by a **single element** in bytes.
   - The **total count** of elements.
   - The **total memory** consumed by the entire array in bytes.

<details>
<summary>💡 Need a hint?</summary>

- Type specification: `dtype='int16'`
- Dimensionality: `arr.ndim`
- Shape: `arr.shape`
- Data type: `arr.dtype`
- Single-item byte size: `arr.itemsize`
- Total elements: `arr.size`
- Total array bytes: `arr.nbytes` (or `arr.itemsize * arr.size`)
</details>

Create a 1-dimensional array

```python
import numpy as np

# 1D Array
a = np.array( [1, 2, 3] , dtype='int16' )
a
```

Create a 2-dimensional array

```python
#2D Array
b = np.array( [ [1, 2, 3] , [4, 5, 6] ] )
b
```

Number of dimensions

```python
print(a.ndim)

print(b.ndim)
```

Shape of the array

```python
print( a.shape )

print( b.shape )
```

Find the Data Type

```python
print( a.dtype )

print( b.dtype )
```

Single Element Memory Size

```python
print( a.itemsize )

print( b.itemsize )
```

Number of elements in the array

```python
print( a.size )

print( b.size )
```

Total memory consumed by the entire array

```python
print( a.itemsize * a.size )

print( b.itemsize * b.size )
```
