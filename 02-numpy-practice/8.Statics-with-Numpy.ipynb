1. Given `mat = np.array([[1, 2, 3], [4, 5, 6]])`:
   - Find the absolute minimum value across the entire matrix.
   - Find the minimum values calculated **column-wise**.
   - Find the maximum values calculated **row-wise**.
   - Calculate the sum of all elements.

```python
import numpy as np

mat = np.array([[1, 2, 3], [4, 5, 6]])

# Minimum Value
a1 = np.min(mat)
print('Min value is: ', a1)

# Maximum Value Columnwise
a2 = np.max(mat, axis=0)
print('Max value columnwise is: ', a2)

# Maximum Value Rowwise
a3 = np.max(mat, axis=1)
print('Max value rowwise is: ', a3)

# Sum Value
a4 = np.sum(mat)
print('Sum value is: ', a4)
```

2. Given `vals = np.array([-3, -2, -1, 0, 1, 2, 3, 4, 5])`:
   - Compute square roots (observing where `nan` appears).
   - Compute absolute values.
   - Compute exponential values ($e^x$).
   - Compute natural logarithms.

```python
vals = np.array([-3, -2, -1, 0, 1, 2, 3, 4, 5])

#Square Root Values
b1 = np.sqrt(vals)
print('Squaroot values: ', b1)

#Absolute Values
b2 = np.abs(vals)
print('Absolute values: ', b2)

#Exponential Values
b3 = np.exp2(vals)
print('Exponential values: ', b3)

#Log Values
b4 = np.log(vals)
print('Log values: ', b4)
```
