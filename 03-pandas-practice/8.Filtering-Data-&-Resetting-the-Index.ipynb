### Exercise 8: Filtering Data & Resetting the Index

**Requirements:**
1. Find all Pokémon that are **BOTH** `'Grass'` Type 1 **AND** `'Poison'` Type 2.
2. Find all Pokémon that are **EITHER** `'Grass'` Type 1 **OR** `'Poison'` Type 2.
3. Find all Pokémon that are **NOT** Grass and **NOT** Poison.
4. Take that filtered result (not Grass, not Poison), reset the index back to 0, 1, 2..., and remove the old index column.

<details>
<summary>💡 Need a hint?</summary>

- AND filter: `df.loc[(df['Type 1'] == 'Grass') & (df['Type 2'] == 'Poison')]`
- OR filter: `df.loc[(df['Type 1'] == 'Grass') | (df['Type 2'] == 'Poison')]`
- NOT filter: `df.loc[~((df['Type 1'] == 'Grass') | (df['Type 2'] == 'Poison'))]`
- Reset index cleanly: `.reset_index(drop=True)`
</details>

----

Import TXT dataset as Dataframe

```python
import pandas as pd

df = pd.read_csv('./data/pokemon_data.txt', delimiter='\t')

df.head()
```

1. Find all Pokémon that are **BOTH** `'Grass'` Type 1 **AND** `'Poison'` Type 2.

```python
df.loc[ (df['Type 1']=='Grass') & (df['Type 2']=='Poison') ]
```

2. Find all Pokémon that are **EITHER** `'Grass'` Type 1 **OR** `'Poison'` Type 2.

```python
df.loc[ (df['Type 1']=='Grass') | (df['Type 2']=='Poison') ]
```

3. Find all Pokémon that are **NOT** Grass and **NOT** Poison.

```python
df.loc[ ~((df['Type 1']=='Grass') & (df['Type 2']=='Poison')) ]
```

4. Take that filtered result (not Grass, not Poison), reset the index back to 0, 1, 2..., and remove the old index column.

```python
df.loc[ ~((df['Type 1']=='Grass') & (df['Type 2']=='Poison')) ] .reset_index(drop=True)
```
