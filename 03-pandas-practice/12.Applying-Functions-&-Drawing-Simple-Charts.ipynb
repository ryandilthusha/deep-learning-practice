Import TXT dataset as Dataframe

```python
import pandas as pd

df = pd.read_csv('./data/pokemon_data.txt', delimiter='\t')

df.head()
```

1. Write a Python function `multiply_100(x)` that multiplies any number by 100.

```python
def multiply_100(x):
    return x*100

print(multiply_100(2))
```

2. Apply this function to the `'Attack'` column.

```python
df['Attack'].head()
```

```python
df['Attack'].apply(func=multiply_100)
```

3. Apply this function to both `'Attack'` and `'Defense'` columns together.

```python
df[['Attack','Defense']].head()
```

```python
df[['Attack','Defense']].apply(func=multiply_100)
```

4. Make a simple histogram chart for column `'A'` from a small random table:
   - Give it a title: `'Frequency Distribution of Column A'`.
   - Set the size to `(10, 6)`.
   - Make the color `'skyblue'`.
   - Set `bins=15`.
   - Turn on the grid and show the legend.

```python
# Import MAtplotlib
%matplotlib inline
```

```python
# Create a simple Dataframe with Columns A, B, C
import numpy as np

a = np.random.randint(2,10,(4,3))

a
```

```python
df_chart = pd.DataFrame(data=a, columns=['A','B','C'])

df_chart
```

```python
df_chart['A'].plot(
    kind='hist',
    title='Freq Dist of A',
    figsize=(10,6),
    color='red',
    bins=15,
    grid=True,
    legend=True

)
```
