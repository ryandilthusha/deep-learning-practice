### Exercise 9: Searching Text, Using Regex, and Changing Values

**Requirements:**
1. Find all Pokémon that have `'Mega'` anywhere inside their name.
2. Using regular expressions (ignore uppercase and lowercase):
   - Find Pokémon where `'Type 1'` is either `'fire'` or `'grass'`.
   - Find Pokémon whose names contain `'pi'` anywhere.
   - Find Pokémon whose names **start** with `'pi'`.
3. Updating table values:
   - In the `'Type 1'` column, change every `'Fire'` to `'FLAME'`.
   - If a Pokémon has `'HP'` greater than 50, set its `'Generation'` to `999` and `'Legendary'` to `True`.

<details>
<summary>💡 Need a hint?</summary>

- Text search: `df.loc[df['Name'].str.contains('Mega')]`
- Regex search: `import re; df.loc[df['Name'].str.contains('^pi', flags=re.I, regex=True)]`
- Change single column: `df.loc[df['Type 1'] == 'Fire', 'Type 1'] = 'FLAME'`
- Change multiple columns: `df.loc[df['HP'] > 50, ['Generation', 'Legendary']] = [999, True]`
</details>

Import TXT dataset as Dataframe

```python
import pandas as pd

df = pd.read_csv('./data/pokemon_data.txt', delimiter='\t')

df.head()
```

1. Find all Pokémon that have `'Mega'` anywhere inside their name.

```python
df.loc[df['Name'].str.contains('Mega')]
```

2. Using regular expressions (ignore uppercase and lowercase):
   - Find Pokémon where `'Type 1'` is either `'fire'` or `'grass'`.

```python
df.loc[df['Type 1'].str.contains('fire|grass', case=False, regex=True)]
```

2. Using regular expressions (ignore uppercase and lowercase):
   - Find Pokémon whose names contain `'pi'` anywhere.

```python
df.loc[df['Name'].str.contains('pi', case=False, regex=True)]
```

2. Using regular expressions (ignore uppercase and lowercase):
   - Find Pokémon whose names **start** with `'pi'`.

```python
df.loc[df['Name'].str.contains('^pi', case=False, regex=True)]
```

3. Updating table values:
   - In the `'Type 1'` column, change every `'Fire'` to `'FLAME'`.

```python
df.loc[df['Type 1'].str.contains('fire', case=False, regex=True), ['Type 1']] = 'FLAME'

df.loc[df['Type 1'].str.contains('flame', case=False, regex=True)]
```

3. Updating table values:
   - If a Pokémon has `'HP'` greater than 50, set its `'Generation'` to `999` and `'Legendary'` to `True`.

```python
df.loc[df['HP']>50, ['Generation','Legendary']]
```
