### Exercise 5: Sorting Values

**Requirements:**
1. Sort the table alphabetically (A to Z) by `'Name'`.
2. Sort the table by both `'Name'` and `'HP'` in reverse order (Z to A, highest to lowest).
3. Sort `'Name'` in alphabetical order (A to Z), but if two names are the same, show the higher `'HP'` first.

<details>
<summary>💡 Need a hint?</summary>

- Single column sort: `df.sort_values('Name')`
- Both descending: `df.sort_values(['Name', 'HP'], ascending=False)`
- Mixed sorting: `df.sort_values(['Name', 'HP'], ascending=[True, False])`
</details>

---

Import TXT data file as Data Frame

```python
import pandas as pd

df = pd.read_csv('./data/pokemon_data.txt' , delimiter='\t')

df.head()
```

1. Sort the table alphabetically (A to Z) by `'Name'`.

```python
df.sort_values(by= 'Name')
```

2. Sort the table by both `'Name'` and `'HP'` in reverse order (Z to A, highest to lowest).

```python
df.sort_values(by=['Name', 'HP'], ascending=False)
```

3. Sort `'Name'` in alphabetical order (A to Z), but if two names are the same, show the higher `'HP'` first.

```python
df.sort_values(by=['Name','HP'], ascending=[True,False])
```
