Import TXT and show as Dataframe

```python
import pandas as pd

df = pd.read_csv('./data/pokemon_data.txt' , delimiter='\t')

df.head()
```

1. Using regular brackets `[]`, show only the `'Name'` column for the first 5 rows.

```python
df['Name'] .head(5)
```

2. Using regular brackets `[]`, show `'Name'`, `'Type 1'`, and `'Type 2'` for the first 5 rows together.

```python
df[['Name','Type 1','Type 2']] .head()
```

3. Using `.iloc` (by row/column numbers):
   - Get rows from position 2 to 6.

```python
df.iloc[ 1:7 , : ]
```

3. Using `.iloc` (by row/column numbers):
   - Get the single value at row position 6 and column position 2.

```python
df.iloc[ 7 , 2 ]
```

4. Using `.loc` (by column names and labels):
   - Get `'Name'` and `'HP'` for row labels 0, 1, and 2.

```python
df.loc[  0:2 , ['Name','HP'] ]
```

4. Using `.loc` (by column names and labels):
   - Get only the `'Name'` and `'Legendary'` columns for Pokémon where `'Legendary'` is `True`.

```python
df.loc[ df['Legendary'] == True , ['Name', 'Legendary']  ]  .head()
```

df.loc[df['Legendary'], ['Name', 'Legendary']]

4. Using `.loc` (by column names and labels):
   - Get the first 5 Pokémon that have `'Attack'` greater than 150.

```python
df.loc[df['Attack'] > 150].head(5)
```
