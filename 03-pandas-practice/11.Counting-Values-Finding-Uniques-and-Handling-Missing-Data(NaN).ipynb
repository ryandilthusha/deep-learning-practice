4. Using this small sample table:
   ```python
   my_data = {
       "A": [1, 2, 3],
       "B": [np.nan, 5, np.nan],
       "C": [7, 8, np.nan]
   }
   df_new = pd.DataFrame(data=my_data)
   ```
   - Drop all columns that have any `NaN` values.
   - Drop all rows that have any `NaN` values.
   - Keep only rows that have at least 2 real (non-NaN) values.
   - Fill all `NaN` values with the text `'Ryan'`.

````python
import numpy as np

a = np.array([[1,2,3],[np.nan,5,np.nan],[7,8,np.nan]])

a
````

````python
df_new = pd.DataFrame(data=a, columns=["A","B","C"])

df_new
````

   - Drop all columns that have any `NaN` values.

````python
df_new.dropna(axis=1)
````

   - Drop all rows that have any `NaN` values.

````python
df_new.dropna(axis=0)
````

   - Keep only rows that have at least 2 real (non-NaN) values.

````python
df_new.dropna(axis=0, thresh=2)
````

   - Fill all `NaN` values with the text `'Ryan'`.

````python
df_new.fillna('Ryan')
````
